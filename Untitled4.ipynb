```python
import os
from pathlib import Path

from google.colab import userdata

from langchain_google_genai import (
    ChatGoogleGenerativeAI,
    GoogleGenerativeAIEmbeddings
)

from langchain_community.document_loaders import PyPDFLoader
from langchain_community.vectorstores import FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
os.environ["GOOGLE_API_KEY"] = userdata.get("GOOGLE_API_KEY")
```

```python
from google.colab import files

uploaded = files.upload()

PDF_PATH = next(iter(uploaded.keys()))

print("Loaded file:", PDF_PATH)
```

```python
loader = PyPDFLoader(PDF_PATH)

documents = loader.load()

print("Number of pages:", len(documents))
```

```python
print(documents[0].page_content[:1000])
print(documents[0].metadata)
```

```python
for doc in documents:
    doc.metadata["document_name"] = Path(PDF_PATH).name
```

```python
splitter = RecursiveCharacterTextSplitter(
    chunk_size=700,
    chunk_overlap=100,
    separators=[
        "\n\n",
        "\n",
        ". ",
        " ",
        ""
    ]
)

chunks = splitter.split_documents(documents)

print("Number of chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks):
    chunk.metadata["chunk_id"] = i
```

```python
for i, chunk in enumerate(chunks[:5]):

    print(f"\nChunk {i + 1}")

    print("Text:")
    print(chunk.page_content[:500])

    print("\nMetadata:")
    print(chunk.metadata)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
FAISS_PATH = "faiss_index"
```

```python
if os.path.exists(FAISS_PATH):

    print("Loading existing FAISS index...")

    vectorstore = FAISS.load_local(
        FAISS_PATH,
        embeddings,
        allow_dangerous_deserialization=True
    )

else:

    print("Creating new FAISS index...")

    vectorstore = FAISS.from_documents(
        chunks,
        embeddings
    )

    vectorstore.save_local(FAISS_PATH)

    print("FAISS index saved.")
```

```python
retriever = vectorstore.as_retriever(
    search_type="mmr",
    search_kwargs={
        "k": 4,
        "fetch_k": 10
    }
)
```

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-2.5-flash",
    temperature=0
)
```

```python
RAG_PROMPT = """
You are a helpful university assistant.

Answer the user's question using ONLY the provided context.

Rules:
1. Do not use outside knowledge.
2. Do not make up information.
3. If the answer is not present in the context, reply:
   "I don't know based on the provided documents."
4. Keep the answer clear and concise.
5. Mention the source and page number when available.
6. Do not invent citations.

Context:
{context}

Question:
{question}

Answer:
"""
```

```python
def format_context(docs):

    formatted = []

    for i, doc in enumerate(docs):

        source = doc.metadata.get("source", "Unknown")
        page = doc.metadata.get("page", "Unknown")
        chunk_id = doc.metadata.get("chunk_id", "Unknown")

        formatted.append(
            f"""
--- Document {i + 1} ---
Source: {source}
Page: {page}
Chunk ID: {chunk_id}

{doc.page_content}
"""
        )

    return "\n".join(formatted)
```

```python
def retrieve_with_debug(question):

    print("\n" + "=" * 70)
    print("USER QUESTION")
    print("=" * 70)

    print(question)

    docs = retriever.invoke(question)

    print("\n" + "=" * 70)
    print("RETRIEVED CHUNKS")
    print("=" * 70)

    for i, doc in enumerate(docs):

        print(f"\n--- Retrieved Chunk {i + 1} ---")

        print("Source:",
              doc.metadata.get("source"))

        print("Page:",
              doc.metadata.get("page"))

        print("Chunk ID:",
              doc.metadata.get("chunk_id"))

        print("\nContent:")
        print(doc.page_content)

    context = format_context(docs)

    print("\n" + "=" * 70)
    print("CONTEXT SENT TO LLM")
    print("=" * 70)

    print(context)

    return docs, context
```

```python
def ask_rag(question):

    docs, context = retrieve_with_debug(question)

    prompt = RAG_PROMPT.format(
        context=context,
        question=question
    )

    print("\n" + "=" * 70)
    print("PROMPT SENT TO LLM")
    print("=" * 70)

    print(prompt)

    response = llm.invoke(prompt)

    print("\n" + "=" * 70)
    print("FINAL ANSWER")
    print("=" * 70)

    print(response.content)

    print("\n" + "=" * 70)
    print("SOURCES")
    print("=" * 70)

    seen = set()

    for doc in docs:

        source = doc.metadata.get("source", "Unknown")
        page = doc.metadata.get("page", "Unknown")

        citation = (source, page)

        if citation not in seen:

            print(f"- {source}, Page {page}")

            seen.add(citation)

    return response.content
```

```python
answer = ask_rag(
    "What is the population of Japan?"
)
```

```python
answer = ask_rag(
    "What is the main difference between C++ set and unordered_set in terms of underlying implementation and average time complexity"
)
```
