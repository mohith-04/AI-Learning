```python
from datasets import load_dataset
from transformers import AutoTokenizer

# load dataset
dataset = load_dataset("imdb")

# take small subset
small_dataset = dataset["train"].select(
    range(5)
)

# tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

# preprocessing function
def preprocess(example):

    return tokenizer(
        example["text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

# tokenize
tokenized_dataset = small_dataset.map(
    preprocess
)

print(tokenized_dataset[0])
```

```python
from datasets import load_dataset
from transformers import AutoTokenizer

# load dataset
dataset = load_dataset("stanfordnlp/imdb")

# small subset
small_dataset = dataset["train"].select(
    range(5)
)

# tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

# preprocessing function
def preprocess(example):

    return tokenizer(
        example["text"],
        truncation=True,
        padding="max_length",
        max_length=128
    )

# tokenize dataset
tokenized_dataset = small_dataset.map(
    preprocess
)

# print first example
print(
    tokenized_dataset[0]
)
```
