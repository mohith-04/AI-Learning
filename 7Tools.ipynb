```python
def calculator(expression):

    try:
        return eval(expression)

    except Exception:
        return "Invalid calculation"


def get_weather(city):

    weather_data = {

        "hyderabad":
        "31°C, Sunny",

        "vijayawada":
        "34°C, Cloudy",

        "visakhapatnam":
        "30°C, Rainy"
    }

    return weather_data.get(
        city.lower(),
        "Weather not found"
    )


def get_student(student_id):

    students = {

        101: {
            "name": "Mohith",
            "course":
            "AI Engineering"
        },

        102: {
            "name": "Rahul",
            "course":
            "Data Science"
        }
    }

    student = students.get(
        student_id
    )

    if student:

        return student

    return "Student not found"


print(
    calculator("5 * 8")
)

print(
    get_weather(
        "Hyderabad"
    )
)

print(
    get_student(101)
)



# 1st create 
```
