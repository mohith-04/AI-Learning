```python
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

import torch

model_name = (
    "distilbert-base-uncased-finetuned-sst-2-english"
)

tokenizer = AutoTokenizer.from_pretrained(
    model_name
)

model = AutoModelForSequenceClassification.from_pretrained(
    model_name
)

text = "This course is amazing"

inputs = tokenizer(
    text,
    return_tensors="pt"
)

outputs = model(**inputs)

# logits
logits = outputs.logits
print("Logits:", logits)

# probabilities
probabilities = torch.softmax(
    logits,
    dim=1
)

print(
    "Probabilities:",
    probabilities
)

# predicted class
prediction = torch.argmax(
    logits,
    dim=1
)

label_map = {
    0: "Negative",
    1: "Positive"
}

print(
    "Prediction:",
    label_map[
        prediction.item()
    ]
)
```

Embeddings

```python
from transformers import (
    AutoTokenizer,
    AutoModel
)

import torch
import torch.nn.functional as F

tokenizer = AutoTokenizer.from_pretrained(
    "bert-base-uncased"
)

model = AutoModel.from_pretrained(
    "bert-base-uncased"
)

sentences = [
    "I love AI",
    "I enjoy artificial intelligence"
]

inputs = tokenizer(
    sentences,
    padding=True,
    truncation=True,
    return_tensors="pt"
)

outputs = model(**inputs)

embeddings = outputs.last_hidden_state.mean(dim=1)

similarity = F.cosine_similarity(
    embeddings[0].unsqueeze(0),
    embeddings[1].unsqueeze(0)
)

print(similarity)
```

```python
#temperature means randomness/creativity
#low temp more focused more predictable less creative and best for facts coding summaries
#high temp means more random more creative less predictaable best for stories, brainstroming and creative writing
from transformers import (
    AutoTokenizer,
    AutoModelForCausalLM
)

model_name = "gpt2"

tokenizer = AutoTokenizer.from_pretrained(
    model_name
)

model = AutoModelForCausalLM.from_pretrained(
    model_name
)

prompt = "The future of robotics"

inputs = tokenizer(
    prompt,
    return_tensors="pt"
)

outputs = model.generate(
    **inputs,
    max_new_tokens=50,
    temperature=0.8,
    do_sample=True
)

generated_text = tokenizer.decode(
    outputs[0],
    skip_special_tokens=True
)

print(generated_text)
```

movie review


```python
from transformers import (
    AutoTokenizer,
    AutoModelForSequenceClassification
)

import torch

# model name
model_name = (
    "distilbert-base-uncased-finetuned-sst-2-english"
)

# load tokenizer
tokenizer = AutoTokenizer.from_pretrained(
    model_name
)

# load model
model = AutoModelForSequenceClassification.from_pretrained(
    model_name
)

# user input
review = input("Enter movie review: ")

# tokenize text
inputs = tokenizer(
    review,
    return_tensors="pt",
    truncation=True,
    padding=True
)

# inference
with torch.no_grad():
    outputs = model(**inputs)

# logits
logits = outputs.logits

# probabilities
probabilities = torch.softmax(
    logits,
    dim=1
)

# predicted class
prediction = torch.argmax(
    probabilities,
    dim=1
).item()

# label mapping
label_map = {
    0: "Negative",
    1: "Positive"
}

label = label_map[prediction]

confidence = (
    probabilities[0][prediction].item()
    * 100
)

# final output
print("\nPrediction:")
print(
    f"{label} ({confidence:.2f}%)"
)
```

Fake News Detector

```python
from transformers import AutoTokenizer
from transformers import AutoModelForSequenceClassification
import torch

# Model name
model_name = "hamzab/roberta-fake-news-classification"

# Load tokenizer and model
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)

# Input news text
news = input("Enter news text: ")

# Tokenize input text
inputs = tokenizer(
    news,
    return_tensors="pt",
    truncation=True,
    padding=True,
    max_length=512
)

# Disable gradient calculation for inference
with torch.no_grad():
    outputs = model(**inputs)

# Get logits
logits = outputs.logits

# Convert logits to probabilities
probabilities = torch.softmax(logits, dim=1)

# Get predicted class index
predicted_class = torch.argmax(probabilities, dim=1).item()

# Labels
labels = ["FAKE", "REAL"]

# Confidence score
confidence = probabilities[0][predicted_class].item()

# Print result
print("\nPrediction:", labels[predicted_class])
print("Confidence:", round(confidence * 100, 2), "%")
```
