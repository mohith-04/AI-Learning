```python
import torch
import torch.nn as nn

class simpleNN(nn.Module):
  def __init__(self):
    super().__init__()
    self.linear=nn.Linear(1,1)
  def forward(self,x):
    return self.linear(x)
model=simpleNN()
x=torch.tensor([[5.0]])
out=model(x)
print(out)
```

```python
import torch
import torch.nn as nn

class simpleNN(nn.Module):
  def __init__(self):
    super().__init__()
    self.hidden=nn.Linear(2,4)
    self.output=nn.Linear(4,1)
  def forward(self,x):
    x=self.hidden(x)
    x=self.output(x)
    return x
model=simpleNN()
x=torch.tensor([[5.0,4.0]])
out=model(x)
print(out)
```

```python
import torch
import torch.nn as nn

model=nn.Linear(1,1)
correction=nn.MSELoss()
optimizer=torch.optim.SGD(model.parameters(),lr=0.1)
x_train=torch.tensor([[1.0],[2.0],[3.0]])
y_train=torch.tensor([[2.0],[4.0],[6.0]])

epochs=100
for epoch in range(epochs):
  output=model(x_train)
  loss=correction(output,y_train)
  optimizer.zero_grad()
  loss.backward()
  optimizer.step()

  print(loss.item())
```

```python
x_test=torch.tensor([[5.0]])
pre=model(x_test)
print(pre.item())
```
